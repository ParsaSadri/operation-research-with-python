$$
\text{Choose decision variables}
$$

to

$$
\text{maximize/minimize an objective function}
$$

subject to

$$
\text{a set of constraints}
$$

A small workshop produces two products: **A** and **B**.

- Each unit of product **A** generates a profit of **30**.
- Each unit of product **B** generates a profit of **20**.

Production requires two limited resources:

- **Machine time**
- **Raw material**

Resource requirements per unit are:

- Product **A** needs:
  - 2 hours of machine time
  - 1 unit of raw material
- Product **B** needs:
  - 1 hour of machine time
  - 1 unit of raw material

Available resources:

- Total machine time available: **100**
- Total raw material available: **80**

---

## Question

How many units of products **A** and **B** should be produced in order to **maximize total profit**?

  <h2 style="margin: 0; color: #1f3b73; font-size: 22px;">
      2. Understanding the Model Components
  </h2>
</div>


Before writing any code, we should identify the elements of the optimization model.

---

## 2.1 Parameters (Known Data)

These values are given in advance and are **not** chosen by us.

Examples in this problem:

- profit of each product
- machine time requirement
- raw material requirement
- total available machine time
- total available raw material

---

## 2.2 Decision Variables

These are the unknowns that the model must determine.

In this problem:

- How many units of product **A** should we produce?
- How many units of product **B** should we produce?

Let:

$$
x_A = \text{number of units of product A}
$$

$$
x_B = \text{number of units of product B}
$$

---

## 2.3 Objective Function

The objective tells us what we want to optimize.

Here, we want to **maximize total profit**.

Profit from product A:

$$
30x_A
$$

Profit from product B:

$$
20x_B
$$

So the total profit is:

$$
\max Z = 30x_A + 20x_B
$$

---

## 2.4 Constraints

Constraints represent the real-world limitations.

### Machine time constraint

Each unit of A uses 2 machine hours, and each unit of B uses 1 machine hour.

Since total machine time is at most 100:

$$
2x_A + x_B \leq 100
$$

### Raw material constraint

Each unit of A uses 1 unit of raw material, and each unit of B also uses 1 unit.

Since total raw material is at most 80:

$$
x_A + x_B \leq 80
$$

### Non-negativity constraints

We cannot produce a negative number of products:

$$
x_A \geq 0,\qquad x_B \geq 0
$$

  <h2 style="margin: 0; color: #1f3b73; font-size: 22px;">
      3. Mathematical Formulation
  </h2>
</div>


The complete mathematical model is:

$$
\max Z = 30x_A + 20x_B
$$

subject to

$$
2x_A + x_B \leq 100
$$

$$
x_A + x_B \leq 80
$$

$$
x_A \geq 0,\qquad x_B \geq 0
$$

---

## Interpretation

- The model chooses values for $x_A$ and $x_B$.
- These values must satisfy all constraints.
- Among all feasible choices, we want the one with the highest profit.

### Import main packages

```python
import numpy as np
import pandas as pd
```

### Variables and Constraints

```python
# Define the set of products
products = ['A', 'B']

# Profit per unit
profit = {'A': 30,'B': 20}

# Resource requirements per unit
machine_req = {'A': 2,'B': 1}
material_req = {'A': 1,'B': 1}

# Resource capacities
capacity = {'machine': 100,'material': 80}
```

```python
print("Products:", products)
print("Profit:", profit)
print("Machine requirements:", machine_req)
print("Material requirements:", material_req)
print("Capacities:", capacity)
```

## Why dictionaries?

Dictionaries are very useful in optimization modeling because they let us access data by name.

For example:

- `profit['A']` gives the profit of product A
- `machine_req['B']` gives the machine requirement of product B

This is more readable and scalable than using many separate variables.


```python
# Organize the product data into a DataFrame
df_products = pd.DataFrame({'product': products,
                            'profit': [profit[p] for p in products],
                            'machine_req': [machine_req[p] for p in products],
                            'material_req': [material_req[p] for p in products]})
df_products
```

    6. Manual Evaluation of Candidate Solutions

Before using an optimization solver, we can check a few candidate solutions manually.

This helps us understand two important ideas:

1. **Feasibility**: Does a solution satisfy all constraints?
2. **Objective value**: How good is the solution?

A candidate solution may have high profit but still be infeasible.


```python
candidates = [
    {'A': 0,  'B': 0},
    {'A': 20, 'B': 40},
    {'A': 40, 'B': 20},
    {'A': 50, 'B': 0},
    {'A': 10, 'B': 70},
    {'A': 30, 'B': 50}
]

candidates
```

## Define a function to Evaluate Solution

```python
def evaluate_solution(solution, profit, machine_req, material_req, capacity):
    machine_used = sum(machine_req[p] * solution[p] for p in solution)
    material_used = sum(material_req[p] * solution[p] for p in solution)
    total_profit = sum(profit[p] * solution[p] for p in solution)
    
    feasible = (machine_used <= capacity['machine'] and material_used <= capacity['material'] and all(solution[p] >= 0 for p in solution))
    
    return {'A': solution['A'],'B': solution['B'],'machine_used': machine_used,'material_used': material_used,'total_profit': total_profit,'feasible': feasible}
```

```python
results = []

for sol in candidates:
    summary = evaluate_solution(solution=sol, profit=profit, machine_req=machine_req, material_req=material_req, capacity=capacity)
    results.append(summary)

df_results = pd.DataFrame(results)
df_results
```

## Interpreting the table

For each candidate solution, we check:

- how much machine time is used,
- how much raw material is used,
- the total profit,
- whether the solution is feasible.

A feasible solution satisfies all constraints.

Among feasible solutions, a higher objective value is better.


```python
# Show only feasible solutions
df_feasible = df_results[df_results['feasible'] == True].copy()
df_feasible
```

```python
# Sort feasible solutions by profit (highest first)
df_feasible.sort_values(by='total_profit', ascending=False)
```

## Important observation

Checking a few candidate solutions manually is **useful for learning, but it is not enough in general.**

Why?

Because there may be many possible combinations of production quantities.

In real optimization problems, the number of possible solutions can be extremely large.

That is why we need optimization algorithms and solvers.

In this first notebook, however, our main goal is to understand the structure of the model.

    7. Exploring the Solution Space with Random Samples
  </h2>

So far we checked only a few manually defined candidate solutions.

But the possible combinations of production quantities are very large.

To better understand the **feasible region**, we will:

1. Randomly generate many candidate solutions
2. Evaluate feasibility and profit
3. Visualize where feasible solutions tend to appear

This gives us intuition about the **geometry of the problem**.


```python
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
N = 2000
random_solutions = []
for i in range(N):
    sol = {'A': np.random.uniform(0, 80), 'B': np.random.uniform(0, 80)}
    
    summary = evaluate_solution(solution=sol, profit=profit, machine_req=machine_req, material_req=material_req, capacity=capacity)
    random_solutions.append(summary)

df_random = pd.DataFrame(random_solutions)
df_random.head()
```

```python
df_random['feasible'].value_counts()
```

## Scatter Plot of Candidate Solutions

Each point represents one possible production plan:

- x-axis → units of product A
- y-axis → units of product B

Colors indicate feasibility:

- **Green** → feasible
- **Red** → infeasible

This helps us visually understand the feasible region.


```python
plt.figure(figsize=(7,6))

feasible = df_random[df_random['feasible']]
infeasible = df_random[~df_random['feasible']]

plt.scatter(infeasible['A'], infeasible['B'], color='red', alpha=0.3, label='Infeasible')
plt.scatter(feasible['A'], feasible['B'], color='green', alpha=0.5, label='Feasible')

plt.xlabel("Units of Product A")
plt.ylabel("Units of Product B")
plt.title("Feasible vs Infeasible Solutions")
plt.legend()

plt.show()
```

## Visualizing the Real Feasible Region (Polygon)

Before introducing the Simplex algorithm, it is extremely helpful to 
visually show the **feasible polygon** defined by the constraints.

We will plot:

- Constraint line: $2x_A + x_B = 100$  (Machine time)
- Constraint line: $x_A + x_B = 80$    (Material)
- Non-negativity: $x_A >= 0, x_B >= 0$

Then we will shade the feasible area, identify the **corner points**, and
explain why Simplex moves among vertices.


```python
import numpy as np
import matplotlib.pyplot as plt

A_vals = np.linspace(0, 100, 400)
line_machine = 100 - 2*A_vals       # x_B = 100 - 2 x_A
line_material = 80 - A_vals         # x_B = 80 - x_A
```

```python
plt.figure(figsize=(7,6))

plt.plot(A_vals, line_machine, label="2A + B = 100  (Machine)", color='blue')
plt.plot(A_vals, line_material, label="A + B = 80  (Material)", color='orange')

A = np.linspace(0, 80, 400)
B = np.minimum(100 - 2*A, 80 - A)
B[B < 0] = 0 
plt.fill_between(A, B, color='green', alpha=0.3, label="Feasible Region")

corners = np.array([[0, 0], [0, 80], [20, 60], [50, 0]])
plt.scatter(corners[:,0], corners[:,1], color='black', s=60, label="Vertices")

for (x,y) in corners:
    plt.text(x+1, y+1, f"({int(x)},{int(y)})")

plt.xlim(0, 80)
plt.ylim(0, 80)
plt.xlabel("Units of A")
plt.ylabel("Units of B")
plt.title("Feasible Region of the Linear Program")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(5,4))

A_vals = np.linspace(0, 80, 400)
machine_line = 100 - 2*A_vals
material_line = 80 - A_vals

plt.plot(A_vals, machine_line, label="2A + B = 100", color="blue")
plt.plot(A_vals, material_line, label="A + B = 80", color="orange")

feasible = df_random[df_random["feasible"]]
infeasible = df_random[~df_random["feasible"]]

plt.scatter(infeasible["A"], infeasible["B"], color="red", alpha=0.2, s=10, label="Infeasible")
plt.scatter(feasible["A"], feasible["B"], color="green", alpha=0.5, s=12, label="Feasible")

A = np.linspace(0,80,400)
B = np.minimum(100 - 2*A, 80 - A)
B[B < 0] = 0
plt.fill_between(A, B, color="green", alpha=0.15)

corners = np.array([[0,0],[0,80],[20,60],[50,0]])
plt.scatter(corners[:,0], corners[:,1], color="black", s=60)

for x,y in corners:
    plt.text(x+1, y+1, f"({int(x)},{int(y)})")

plt.xlim(0,80)
plt.ylim(0,80)
plt.xlabel("Units of Product A")
plt.ylabel("Units of Product B")
plt.title("Feasible Solutions and Constraint Lines")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

    8. A More Structured Mathematical View
  </h2>
</div>


In many optimization problems, it is useful to think in terms of:

- **sets**
- **parameters**
- **variables**

Even though this example is small, we can already express it in a more structured way.

---

## Set

$$
P = \{A, B\}
$$

where $P$ is the set of products.

---

## Parameters

For each product $p \in P$:

- $c_p$ = profit of product $p$
- $a^{(m)}_p$ = machine requirement of product $p$
- $a^{(r)}_p$ = raw material requirement of product $p$

Capacities:

- $b_m$ = available machine time
- $b_r$ = available raw material

---

## Decision Variables

For each product $p \in P$:

$$
x_p = \text{units of product } p \text{ to produce}
$$

---

## Model

$$
\max \sum_{p \in P} c_p x_p
$$

subject to

$$
\sum_{p \in P} a^{(m)}_p x_p \leq b_m
$$

$$
\sum_{p \in P} a^{(r)}_p x_p \leq b_r
$$

$$
x_p \geq 0 \qquad \forall p \in P
$$

---

This form is more general and becomes very useful when the number of products increases.


```python
df_results['machine_slack'] = capacity['machine'] - df_results['machine_used']
df_results['material_slack'] = capacity['material'] - df_results['material_used']

df_results
```

## Slack

Slack tells us how much unused resource remains.

For example:

- machine slack = available machine time - used machine time
- material slack = available material - used material

If slack is negative, the solution violates that constraint.

Slack is an important concept that we will revisit later in the course.
